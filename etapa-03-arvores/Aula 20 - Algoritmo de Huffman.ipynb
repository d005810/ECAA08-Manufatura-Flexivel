### 1. Contexto Operacional e Mapeamento de Tags ISA-5.1
No ambiente industrial de manufatura discreta, o sistema **SCADA-Core** gerencia a telemetria em tempo real proveniente dos controladores logicos programaveis (CLP Siemens S7-1500) e dos cartoes de remotas distribuidas ao longo da **Celula de Manufatura Flexivel e Triagem Automatica de Pecas (FMS)**.

A supervisao e intertravamento de seguranca da celula englobam 5 variaveis e alarmes primarios dos Setores 100, 200 e 300:
1. `TH-101`: Disparo de Rele Termico / Sobrecarga no Motor da Esteira Principal `M-101` (Setor 100)
2. `LS-101`: Chave de Nivel Baixo / Silo de Tarugos Vazio no Infeed (Setor 100)
3. `ZS-200`: Inconsistencia Optica / Peca Desalinhada no Tunel de Dimensoes e Cor (Setor 200)
4. `FULL-300`: Intertravamento de Troca de Lote / Caixa Cheia com 10 pecas nas Baias `CX-301..303` (Setor 300)
5. `ESD-301`: Botao de Parada de Emergencia (*E-STOP*) Acionado (`HS-301`, Setor 300)

Em transmissoes convencionais nao compactadas, esse status e serializado em um formato fixo de **5 bits por amostra** (`[TH-101, LS-101, ZS-200, FULL-300, ESD-301]`). 

Sob altas taxas de amostragem (*scan rate* de $10\text{ ms}$ a $50\text{ ms}$) em barramentos de campo com largura de banda restrita (redes seriais RS-485 Modbus-RTU em $9600\text{ bps}$, telemetria sem fio LoRaWAN/Zigbee de AGVs ou links IIoT MQTT/OPC-UA), o envio continuo de mensagens identicas e redundantes causa saturacao do canal e aumento da latencia de comunicacao.

Como a linha opera em condicoes normais e estaveis na esmagadora maioria do tempo ($\approx 88\%$), a aplicacao do **Algoritmo de Huffman** comprime sem perdas (*lossless*) os quadros antes da transmissao, atribuindo codigos minimos (1 ou 2 bits) aos estados mais frequentes e codigos mais longos aos alarmes esporadicos.

```python
import heapq
import math
import random
import matplotlib.pyplot as plt
import pandas as pd
import warnings

warnings.filterwarnings('ignore')

# Configuracao para exibicao grafica limpa
%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
print("Modulos importados com sucesso para o SCADA-Core FMS.")
```

## 2. Modelo Probabilistico das Mensagens da Celula FMS

Definimos o vetor de estado binario ISA-5.1:
$$\mathbf{x} = [x_{\text{TH-101}},\, x_{\text{LS-101}},\, x_{\text{ZS-200}},\, x_{\text{FULL-300}},\, x_{\text{ESD-301}}] \in \{0, 1\}^5$$

Consideramos os 6 estados operacionais mutuamente exclusivos da celula de manufatura:
1. `00000`: **Operacao Nominal Estavel** -- Esteira em movimento, pecas sendo alimentadas e triadas sem falhas ($88{,}0\%$).
2. `00010`: **Intertravamento de Troca de Lote** -- Uma das caixas atingiu o limite de 10 pecas (`FULL-300`), pausando o infeed temporariamente ($6{,}0\%$).
3. `01000`: **Nivel Baixo no Silo Infeed** -- Sensor `LS-101` ativo, requisitando chamada de reabastecimento via AGV ($3{,}0\%$).
4. `00100`: **Inconsistencia Optica no Tunel** -- Peca desalinhada ou falha temporaria de leitura nos sensores dimensionais/RGB ($1{,}8\%$).
5. `10000`: **Alarme de Sobrecarga Termica** -- Relativo a atuacao do termistor `TH-101` no motor `M-101` ($1{,}0\%$).
6. `00001`: **Parada de Emergencia (E-STOP)** -- Acionamento do botao cogumelo `HS-301` / `ESD-301` ($0{,}2\%$).

```python
# Tabela de probabilidades de telemetria da Celula de Manufatura Flexivel
state_probabilities = {
    '00000': 0.880,  # Operacao Nominal Estavel (Processo sem alarmes)
    '00010': 0.060,  # Intertravamento de Troca de Lote (FULL-300, 10 pecas)
    '01000': 0.030,  # Silo de Infeed Vazio / Nivel Baixo (LS-101)
    '00100': 0.018,  # Inconsistencia no Tunel Optico (ZS-200)
    '10000': 0.010,  # Sobrecarga Termica no Motor M-101 (TH-101)
    '00001': 0.002   # Parada de Emergencia Acionada (ESD-301)
}

state_descriptions = {
    '00000': 'Operacao Nominal Estavel (Sem alarmes)',
    '00010': 'Intertravamento Troca de Lote (FULL-300)',
    '01000': 'Silo Infeed Vazio / Nivel Baixo (LS-101)',
    '00100': 'Inconsistencia no Tunel Optico (ZS-200)',
    '10000': 'Sobrecarga Termica Motor Esteira (TH-101)',
    '00001': 'Parada de Emergencia Acionada (ESD-301)'
}

df_probs = pd.DataFrame([
    {
        'Vetor de Estado': sym,
        'Condicao Operacional': state_descriptions[sym],
        'Probabilidade': p,
        'Frequencia (%)': f"{p * 100:.1f}%"
    }
    for sym, p in state_probabilities.items()
])

print("--- MATRIZ DE PROBABILIDADES DE TELEMETRIA DA FMS ---")
display(df_probs)
```

## 3. Construcao da Arvore de Huffman e Geracao de Codigos Prefixados

A estrutura `HuffmanNode` implementa o no da arvore binaria, comparavel por peso (*frequencia*) para permitir a ordenacao no *Min-Heap* (`heapq`). A funcao `build_huffman_tree` consome os dois nos de menor peso iterativamente ate formar a raiz.

```python
class HuffmanNode:
    """No da Arvore Binaria de Huffman."""
    def __init__(self, symbol=None, freq=0.0, left=None, right=None):
        self.symbol = symbol
        self.freq = freq
        self.left = left
        self.right = right

    def __lt__(self, other):
        # Sobrecarga de operador para ordenacao na fila de prioridade (Min-Heap)
        return self.freq < other.freq

def build_huffman_tree(probabilities: dict) -> HuffmanNode:
    """
    Constroi a Arvore de Huffman a partir do dicionario de probabilidades.
    Utiliza um Min-Heap para garantir complexidade O(N log N).
    """
    heap = [HuffmanNode(symbol=sym, freq=p) for sym, p in probabilities.items()]
    heapq.heapify(heap)

    while len(heap) > 1:
        # Extrai os dois nos com menor probabilidade acumulada
        left = heapq.heappop(heap)
        right = heapq.heappop(heap)

        # Mescla os dois nos em um no intermediario
        merged = HuffmanNode(freq=left.freq + right.freq, left=left, right=right)
        heapq.heappush(heap, merged)

    return heap[0]

def generate_codes(root: HuffmanNode, current_code: str = "", code_dict: dict = None) -> dict:
    """
    Percorre a arvore binaria recursivamente gerando os codigos livres de prefixo.
    Convencao: ramificacao a esquerda = '0', ramificacao a direita = '1'.
    """
    if code_dict is None:
        code_dict = {}
    if root is not None:
        if root.symbol is not None:
            code_dict[root.symbol] = current_code
        generate_codes(root.left, current_code + "0", code_dict)
        generate_codes(root.right, current_code + "1", code_dict)
    return code_dict

# Construcao da arvore e sintese dos codigos
tree_root = build_huffman_tree(state_probabilities)
huffman_codes = generate_codes(tree_root)

# Criacao de tabela explicativa dos codigos gerados
df_codes = pd.DataFrame([
    {
        'Estado Fixo (5 bits)': sym,
        'Codigo Huffman': huffman_codes[sym],
        'Bits Comprimidos': len(huffman_codes[sym]),
        'Probabilidade': state_probabilities[sym],
        'Economia por Amostra': f"{((5 - len(huffman_codes[sym])) / 5) * 100:.1f}%"
    }
    for sym in sorted(huffman_codes.keys(), key=lambda s: len(huffman_codes[s]))
])

print("\n--- DICIONARIO DE CODIGOS DE HUFFMAN OTIMIZADOS ---")
display(df_codes)
```

## 4. Simulacao de Transmissao SCADA em Tempo Real (10.000 Amostras)

Simulamos uma transmissao industrial contÃ­nua de telemetria contendo **10.000 amostras** consecutivas geradas a partir da distribuicao estocastica de operacao da CÃ©lula FMS.

```python
NUM_SAMPLES = 10000
symbols = list(state_probabilities.keys())
weights = list(state_probabilities.values())

# Fixamos a semente para repetibilidade experimental estrita
random.seed(42)
telemetry_stream = random.choices(symbols, weights=weights, k=NUM_SAMPLES)

# 1. Transmissao bruta sem compressao (formato fixo de 5 bits por amostra)
raw_bitstream = "".join(telemetry_stream)
raw_size_bits = len(raw_bitstream)

# 2. Transmissao compactada via Huffman
compressed_bitstream = "".join([huffman_codes[sym] for sym in telemetry_stream])
compressed_size_bits = len(compressed_bitstream)

print("=" * 65)
print(" RESULTADOS DA TRANSMISSAO TELEMETRICA SCADA (10.000 AMOSTRAS)")
print("=" * 65)
print(f"Total de Amostras Processadas:        {NUM_SAMPLES:,}")
print(f"Volume Bruto Sem Compressao:          {raw_size_bits:,} bits ({raw_size_bits / 8 / 1024:.2f} KB)")
print(f"Volume Comprimido via Huffman:        {compressed_size_bits:,} bits ({compressed_size_bits / 8 / 1024:.2f} KB)")
print(f"Bits Economizados na Rede:            {raw_size_bits - compressed_size_bits:,} bits")
print(f"Taxa Real de Economia de Banda:       {100 * (1 - compressed_size_bits / raw_size_bits):.2f}%")
print("=" * 65)
```

## 5. Algoritmo de Decodificacao DeterminÃ­stica e Validacao Sem Perdas (*Lossless*)

O receptor SCADA decodifica o fluxo serial de bits percorrendo a arvore binaria de Huffman da raiz ate as folhas em complexidade $\mathcal{O}(k)$ por simbolo, onde $k \le 5$ e o comprimento do codigo.

```python
def decode_huffman(bitstream: str, root: HuffmanNode) -> list:
    """
    Decodifica um fluxo serial de bits utilizando a arvore de Huffman.
    Complexidade O(k) por simbolo transmitido (maximo de 5 operacoes por amostra).
    """
    decoded_symbols = []
    current = root
    for bit in bitstream:
        if bit == '0':
            current = current.left
        else:
            current = current.right

        # Ao alcancar um no folha, recupera o simbolo e reseta para a raiz
        if current.symbol is not None:
            decoded_symbols.append(current.symbol)
            current = root

    return decoded_symbols

# Execucao da decodificacao no servidor SCADA
decoded_telemetry = decode_huffman(compressed_bitstream, tree_root)

# Validacao formal de integridade estrita
is_identical = (telemetry_stream == decoded_telemetry)
assert is_identical, "Erro critico: Divergencia na descompressao de telemetria!"

print(f"Auditoria de Integridade da Telemetria:")
print(f"  -> Quantidade de simbolos decodificados: {len(decoded_telemetry):,}")
print(f"  -> Recuperacao Fiel Sem Perdas (Lossless): {'SIM (100% IDENTICO)' if is_identical else 'NAO'}")
```

## 6. Analise Teorica da Informacao e Graficos de Desempenho

Comparamos as metricas teoricas de Shannon com o desempenho empirico simulado:
* **Entropia de Shannon:** $H(S) = -\sum P(s_i) \log_2 P(s_i)$
* **Comprimento Medio da Palavra-Codigo:** $L = \sum P(s_i) \cdot l(s_i)$
* **Eficiencia de Codificacao:** $\eta = \frac{H(S)}{L} \times 100\%$

```python
# Calculo das grandezas teoricas de Shannon
entropy = -sum(p * math.log2(p) for p in state_probabilities.values())
avg_length = sum(state_probabilities[sym] * len(huffman_codes[sym]) for sym in state_probabilities)
efficiency = (entropy / avg_length) * 100
redundancy = 100 - efficiency

print(f"Entropia Teorica de Shannon H(S):       {entropy:.4f} bits/simbolo")
print(f"Comprimento Medio da Palavra L:         {avg_length:.4f} bits/simbolo")
print(f"Formato Tradicional Fixo:               5.0000 bits/simbolo")
print(f"Eficiencia do Codigo de Huffman:        {efficiency:.2f}%")
print(f"Redundancia Residual:                   {redundancy:.2f}%")
print(f"Taxa Teorica de Compressao:             {(1 - avg_length / 5) * 100:.2f}%")

# Painel grafico comparativo de desempenho
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5), dpi=120)

# Grafico 1: Comparacao de Volume de Dados Transmitidos
categories = ['Formato Fixo Bruto\n(Sem Compressao)', 'SCADA Huffman\n(Comprimido)']
values_kb = [raw_size_bits / 8 / 1024, compressed_size_bits / 8 / 1024]
bar_colors = ['#dc2626', '#16a34a']

bars = ax1.bar(categories, values_kb, color=bar_colors, width=0.45, edgecolor='#1e293b', linewidth=1.5)
ax1.set_ylabel('Volume de Dados em TrÃ¢nsito (KBytes)', fontsize=11, fontweight='bold')
ax1.set_title(f'Volume Total Transmitido ({NUM_SAMPLES:,} Amostras)\nEconomia Real de {100*(1 - compressed_size_bits/raw_size_bits):.1f}%', fontsize=12, fontweight='bold', pad=12)

for bar in bars:
    yval = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, yval + 0.12, f'{yval:.2f} KB', ha='center', va='bottom', fontweight='bold', fontsize=10)

# Grafico 2: Relacao Comprimento do Codigo vs. Probabilidade de Ocorrencia
ordered_states = sorted(state_probabilities.keys(), key=lambda s: -state_probabilities[s])
probs = [state_probabilities[s] * 100 for s in ordered_states]
code_lens = [len(huffman_codes[s]) for s in ordered_states]

ax2.bar(ordered_states, code_lens, color='#0284c7', alpha=0.85, width=0.5, edgecolor='#0f172a', label='Comprimento do Codigo (bits)')
ax2_twin = ax2.twinx()
ax2_twin.plot(ordered_states, probs, color='#ea580c', marker='o', linewidth=2.5, markersize=8, label='Frequencia de Ocorrencia (%)')

ax2.set_xlabel('Vetor de Estado [TH, LS, ZS, FULL, ESD]', fontsize=11, fontweight='bold')
ax2.set_ylabel('Comprimento do Codigo (bits)', fontsize=11, fontweight='bold', color='#0284c7')
ax2_twin.set_ylabel('Frequencia Operacional (%)', fontsize=11, fontweight='bold', color='#ea580c')
ax2.set_title('Hierarquia de Huffman: Estados Mais Provaveis Recebem Menor Quantidade de Bits', fontsize=12, fontweight='bold', pad=12)
ax2.set_ylim(0, 6)

plt.tight_layout()
plt.show()
```

## 7. Verificacao Formal e Assercoes Deterministicas de Engenharia

Executamos assercoes deterministicas para certificar que o algoritmo atende a todos os requisitos matematicos e normativos:

```python
# 1. Validacao de unicidade dos codigos gerados
assert len(set(huffman_codes.values())) == len(state_probabilities), "Erro: Codigos de Huffman duplicados detectados!"

# 2. Validacao da propriedade de codigo livre de prefixo (Prefix-Free)
for s1, c1 in huffman_codes.items():
    for s2, c2 in huffman_codes.items():
        if s1 != s2:
            assert not c1.startswith(c2), f"Erro de prefixo: Codigo {c2} e prefixo de {c1}!"

# 3. Validacao dos limites de Shannon (H(S) <= L < H(S) + 1)
assert entropy <= avg_length, f"Violacao do Teorema de Shannon: L ({avg_length}) < H ({entropy})"
assert avg_length < entropy + 1.0, f"Violacao do limite superior: L ({avg_length}) >= H+1 ({entropy + 1})"

# 4. Validacao da taxa de compressao teorica esperada (75.56%)
assert abs((1 - avg_length / 5) * 100 - 75.56) < 0.1, "Erro: Taxa de compressao calculada diverge do valor nominal!"

# 5. Validacao da descompressao perfeita em lote completo
assert len(decoded_telemetry) == NUM_SAMPLES, "Erro: Quantidade de amostras decodificadas invalida!"
assert decoded_telemetry == telemetry_stream, "Erro: Conteudo decodificado diverge da transmissao original!"

print("[OK] Todas as 5 assercoes formais de Teoria da Informacao e Engenharia passaram com 100% de sucesso!")
```
